```python
import numpy as np
import scipy.stats as stats
import pandas as pd
import matplotlib.pyplot as plt
import os
import networkx as nx
from sklearn.preprocessing import StandardScaler
from sklearn.covariance import GraphicalLassoCV
from sklearn.covariance import EmpiricalCovariance
from scipy.stats import pearsonr

np.random.seed(0)
```

```python
# read dataset
df = pd.read_csv('Form_Responses_04-02-2026_final.csv')
print(f"Total participiants: {df.shape[0]}")

# CLEAN COLUMN NAMES (Removes hidden spaces)
df.columns = df.columns.str.strip()
df = df.drop('Email', axis=1)

# Filter out failed attention tests (AT)
print(f"\nFailed AT: {df[df['AT'] != 'Option 4'].shape[0]}")
df = df[df['AT'] == 'Option 4']

# df = df[df['Studies'] == 'Engineering']

# Filter out unvalid ages
print(f"\nUnvalid ages: {df[df['Age'] < 18].shape[0] + df[df['Age'] > 29].shape[0]}")
df = df[df['Age'] >= 18]
df = df[df['Age'] <= 29]

print(f"\nValid participiants: {df.shape[0]}")
```

```python
# Gender mapping
GENDER = {
    'Male': 1,
    'Female': 2,
    'Non Binary': 3
}

# Education mapping (ordinal)
EDUCATION = {
    'Bachelor': 1,
    'Master': 2,
    'PhD': 3,
    'Post Doc': 4
}

# Apply to Gender
df['Gender'] = df['Gender'].astype(str).str.strip().map(GENDER)

# Apply to Education
df['Education'] = df['Education'].astype(str).str.strip().map(EDUCATION)

# Count Gender
print(f"Male: {df[df['Gender'] == 1].shape[0]}")
print(f"Female: {df[df['Gender'] == 2].shape[0]}")
print(f"Non Binary: {df[df['Gender'] == 3].shape[0]}")


# Education counts
print("\nEducation counts:")
print(f"Bachelor: {df[df['Education'] == 1].shape[0]}")
print(f"Master: {df[df['Education'] == 2].shape[0]}")
print(f"PhD: {df[df['Education'] == 3].shape[0]}")
print(f"Post Doc: {df[df['Education'] == 4].shape[0]}")
```

```python
SMA = ['SMA_Salience', 'SMA_Tolerance', 'SMA_Mood_Modification',
       'SMA_Relapse', 'SMA_Withdrawal', 'SMA_Conflict']

FEELINGS = ['Tired', 'Stressed', 'Sad', 'Anxious', 'Frustrated']

# calculate BSMAS_score
df['BSMAS_score'] = df[SMA].sum(axis=1)

# calculate Emotion_score
df['Emotion_score'] = df[FEELINGS].sum(axis=1)

# Save cleaned data
df.to_csv('Form_Responses_04-02-2026_cleaned.csv', index=False)
```
